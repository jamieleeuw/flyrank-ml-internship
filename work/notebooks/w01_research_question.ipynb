## 2. The question: decision, action, cost of a wrong call

Decision improved: which pages in a content portfolio get reviewed first, given a
content team only has capacity to review a limited number of pages per cycle (e.g.
the top 20-50).

Who acts: a content strategist or SEO reviewer working through a prioritized queue.

The action: review the flagged page and decide to refresh, expand, protect, prune,
or monitor it — the model doesn't take action itself, it just orders the queue.

Cost of a wrong call: two directions.
- False positive (flagged page isn't actually a problem): wastes a reviewer's limited
  time — the page is fine, so the review produces no benefit, though the cost is
  bounded to one reviewer-hour.
- False negative (a genuinely declining, high-value page isn't flagged and gets missed):
  more costly, since real organic traffic keeps eroding unnoticed until someone catches
  it another way.
This asymmetry is why precision@K (are the top K recommendations actually good?) matters
more here than plain accuracy — the review team only ever looks at the top of the list.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

# Walk up until we find the repo root (where data/raw lives)
while not os.path.isdir("data/raw") and os.getcwd() != os.path.dirname(os.getcwd()):
    os.chdir("..")

print("Working dir:", os.getcwd())

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

declining_rate = (df["trend_direction"].str.lower() == "down").mean()
stale = (df["days_since_last_update"] >= 180)
visible = (df["impressions_90d"] >= 500)
stale_and_visible_pct = (stale & visible).mean()

print(f"Rows: {len(df):,}")
print(f"Declining rate (trend_direction == 'down'): {declining_rate:.3f}")
print(f"Share of pages that are both stale (>=180d) and visible (>=500 impressions/90d): {stale_and_visible_pct:.3f}")
```

Three numbers that make this lane worth pursuing:
1. 30,000 pages in the starter slice alone; the real warehouse has ~520k content
   items and ~79M daily fact rows — plenty of scale for this to be worth 7 weeks.
2. ~54.2% of pages are currently labeled "declining" (trend_direction == down) — a
   large, well-balanced target class, not a rare-event problem.
3. Only ~0.1% of pages are both stale (>=180 days since update) and still visible
   (>=500 impressions/90d) — meaning the naive hand-rule from Week 1-2 flags a very
   narrow slice of the inventory. That's exactly why a learned model matters here:
   the simple rule is too conservative to surface most of the real opportunity, and
   the verified pipeline results back this up — Precision@50 = 0.240 for the baseline
   rule vs. 0.740 for a random forest on the same client-holdout split.
